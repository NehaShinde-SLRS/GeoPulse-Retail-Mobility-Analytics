# Day 5 – Snowflake Geospatial Data Warehouse

## Project: GeoPulse – Hyper-Local Retail Mobility & Location Intelligence Platform

### Objective
Set up a Snowflake data warehouse to store synthetic GPS mobility,
retail location, competitor, and commercial zone datasets.

### Infrastructure
- Warehouse: GEOPULSE_WH
- Database: GEOPULSE_DB
- Schemas: RAW, STAGING, ANALYTICS
- File format: GEOPULSE_CSV_FORMAT
- Stage: GEOPULSE_STAGE

### Raw tables created
1. GPS_PINGS
2. STORE_LOCATIONS
3. COMPETITOR_LOCATIONS
4. STORE_CATCHMENTS
5. COMMERCIAL_ZONES

The raw tables include GEOGRAPHY columns for spatial analysis.


## Dataset Loading and Validation

Two synthetic mobility datasets were loaded into Snowflake.

| Dataset | Expected Rows | Loaded Rows | Geography Values |
|---|---:|---:|---:|
| GPS_PINGS | 50,000 | 50,000 | 50,000 |
| REALISTIC_URBAN_MOBILITY | 140,000 | 140,000 | 140,000 |
| **Total** | **190,000** | **190,000** | **190,000** |

Both datasets passed row-count and GEOGRAPHY-column validation.

All records are synthetic and anonymized. They do not represent
actual consumer movements.


## Geospatial Implementation

Snowflake's ST_MAKEPOINT function was used to construct geographic points.

Example SQL:

```sql
ST_MAKEPOINT(LONGITUDE, LATITUDE)
```

The coordinate order is longitude first, followed by latitude.

ST_ASWKT(LOCATION) was used to inspect stored geographic values
in Well-Known Text (WKT) format.

Example output:

`POINT(73.851777757 18.499877724)`


## Validation SQL

```sql
SELECT
    COUNT(*) AS TOTAL_ROWS,
    COUNT(LOCATION) AS GEOGRAPHY_ROWS,
    MIN(PING_TIMESTAMP) AS FIRST_PING,
    MAX(PING_TIMESTAMP) AS LAST_PING
FROM GEOPULSE_DB.RAW.GPS_PINGS;

SELECT
    COUNT(*) AS TOTAL_ROWS,
    COUNT(LOCATION) AS GEOGRAPHY_ROWS,
    MIN(PING_TIMESTAMP) AS FIRST_PING,
    MAX(PING_TIMESTAMP) AS LAST_PING
FROM GEOPULSE_DB.RAW.REALISTIC_URBAN_MOBILITY;
```
