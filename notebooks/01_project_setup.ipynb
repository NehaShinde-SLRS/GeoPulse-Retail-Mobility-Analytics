```python

from pathlib import Path
import shutil

SOURCE = Path("/content/drive/MyDrive/GeoPulse Project")
DESTINATION = Path("/content/drive/MyDrive/GeoPulse-GitHub-Repo")

copied = []
skipped = []

for source_file in SOURCE.rglob("*"):
    if not source_file.is_file():
        continue

    relative_path = source_file.relative_to(SOURCE)

    # Ignore temporary Jupyter files
    if ".ipynb_checkpoints" in relative_path.parts:
        continue

    destination_file = DESTINATION / relative_path
    destination_file.parent.mkdir(parents=True, exist_ok=True)

    if destination_file.exists():
        skipped.append(str(relative_path))
    else:
        shutil.copy2(source_file, destination_file)
        copied.append(str(relative_path))

print("Missing files copied:", len(copied))
print("Existing files preserved:", len(skipped))

print("\nNew files:")
for file in copied:
    print("+", file)
```
