```python
import os
os.environ["SPARK_LOCAL_IP"] = "127.0.0.1"

from pyspark.sql import SparkSession
from sedona.spark import SedonaContext

spark = (
    SparkSession.builder
    .appName("GeoPulse-Day7-Catchments")
    .config(
        "spark.jars.packages",
        "org.apache.sedona:sedona-spark-shaded-3.5_2.12:1.7.1,"
        "org.datasyslab:geotools-wrapper:1.7.1-28.5"
    )
    .getOrCreate()
)

sedona = SedonaContext.create(spark)

print("Spark version:", spark.version)
print("Sedona initialized successfully!")
```


## Day 7 – Project Summary

### Objective
Generate 500-meter store catchment areas using Apache Sedona and PySpark.

### Work Completed
1. Loaded 20 synthetic retail store locations.
2. Created spatial point geometries from store coordinates.
3. Transformed geometries from EPSG:4326 to EPSG:32643 (UTM Zone 43N).
4. Generated 500-meter buffer polygons around all stores.
5. Validated polygon geometry and calculated area and perimeter.
6. Exported catchment geometries as WKT to `data/synthetic/store_catchments_500m.csv`.

### Validation Results
- Total stores processed: 20
- Valid catchment polygons: 20
- Buffer radius: 500 meters
- Approximate polygon area: 780,361.29 square meters per store
- Output CRS: EPSG:32643

### Outcome
Successfully generated and exported valid 500-meter catchment polygons for all 20 synthetic retail stores.

**Note:** These are synthetic, radius-based catchment areas for project analysis, not observed customer-derived trade areas.
