```python

# GeoPulse - Day 02
# Synthetic GPS Mobility Data Generator

import pandas as pd
import numpy as np
import random
from datetime import datetime, timedelta
from pathlib import Path

# Reproducibility
np.random.seed(42)
random.seed(42)

# Project settings
CITY = "Pune"
NUM_DEVICES = 1000
PINGS_PER_DEVICE = 50

print("GeoPulse - Day 02")
print("City:", CITY)
print("Synthetic devices:", NUM_DEVICES)
print("Pings per device:", PINGS_PER_DEVICE)
print("Libraries imported successfully!")
```

```python

# Pune geographical boundaries (simulated area)

PUNE_LAT_MIN = 18.40
PUNE_LAT_MAX = 18.65

PUNE_LON_MIN = 73.75
PUNE_LON_MAX = 74.00

# Generate synthetic device IDs
device_ids = [
    f"DEVICE_{i:05d}"
    for i in range(1, NUM_DEVICES + 1)
]

# Generate random GPS coordinates
records = []

for device_id in device_ids:
    for _ in range(PINGS_PER_DEVICE):
        latitude = np.random.uniform(
            PUNE_LAT_MIN, PUNE_LAT_MAX
        )

        longitude = np.random.uniform(
            PUNE_LON_MIN, PUNE_LON_MAX
        )

        records.append({
            "DeviceID": device_id,
            "latitude": latitude,
            "longitude": longitude
        })

# Create DataFrame
gps_df = pd.DataFrame(records)

print("GPS coordinates generated successfully!")
print("Total records:", len(gps_df))
print("\nFirst 5 records:")
display(gps_df.head())
```

```python

# Define the starting date
START_DATE = pd.Timestamp("2026-11-17 00:00:00")

# Generate timestamps across 7 days
TOTAL_SECONDS = 7 * 24 * 60 * 60

random_seconds = np.random.randint(
    0,
    TOTAL_SECONDS,
    size=len(gps_df)
)

gps_df["timestamp"] = (
    START_DATE +
    pd.to_timedelta(random_seconds, unit="s")
)

# Sort records by device and timestamp
gps_df = (
    gps_df.sort_values(["DeviceID", "timestamp"])
    .reset_index(drop=True)
)

print("Timestamps generated successfully!")
print("Starting timestamp:", gps_df["timestamp"].min())
print("Ending timestamp:", gps_df["timestamp"].max())

print("\nFirst 5 records:")
display(gps_df.head())
```

```python

# GeoPulse - Dataset Validation

print("========== DATASET VALIDATION ==========")

print("Total records:", len(gps_df))
print("Unique devices:", gps_df["DeviceID"].nunique())

print("\nColumn names:")
print(gps_df.columns.tolist())

print("\nMissing values:")
print(gps_df.isnull().sum())

print("\nLatitude valid:",
      gps_df["latitude"].between(18.40, 18.65).all())

print("Longitude valid:",
      gps_df["longitude"].between(73.75, 74.00).all())

print("Timestamp data type:",
      pd.api.types.is_datetime64_any_dtype(gps_df["timestamp"]))

print("Chronological order:",
      gps_df.groupby("DeviceID")["timestamp"]
      .apply(lambda x: x.is_monotonic_increasing)
      .all())

print("\n========== VALIDATION COMPLETE ==========")
```
