# Day 4: Retail Location and Zoning Datasets
GeoPulse – Hyper-Local Retail Mobility & Location Intelligence Platform

All locations and attributes in this notebook are synthetic and illustrative.

```
import pandas as pd
import numpy as np
import random
import os

np.random.seed(42)
random.seed(42)

BASE_PATH = "/content/gdrive/MyDrive/GeoPulse-GitHub-Repo"
SYNTHETIC_PATH = os.path.join(BASE_PATH, "data", "synthetic")
```

## 1. Synthetic Retail Store Locations

```
store_df = pd.read_csv(os.path.join(SYNTHETIC_PATH, "store_locations.csv"))
display(store_df.head())
print("Total stores:", len(store_df))
```

## 2. Synthetic Competitor Locations

```
competitor_df = pd.read_csv(os.path.join(SYNTHETIC_PATH, "competitor_locations.csv"))
display(competitor_df.head())
print("Total competitors:", len(competitor_df))
```

## 3. Synthetic Commercial Zones

```
zone_df = pd.read_csv(os.path.join(SYNTHETIC_PATH, "commercial_zones.csv"))
display(zone_df)
print("Total zones:", len(zone_df))
```

## 4. Synthetic Store Catchments

```
catchment_df = pd.read_csv(os.path.join(SYNTHETIC_PATH, "store_catchments.csv"))
display(catchment_df.head())
print("Total catchments:", len(catchment_df))
```

## 5. Dataset Validation

```
datasets = {
    "Stores": (store_df, "StoreID"),
    "Competitors": (competitor_df, "CompetitorID"),
    "Commercial Zones": (zone_df, "ZoneID"),
    "Catchments": (catchment_df, "CatchmentID")
}

for name, (df, id_column) in datasets.items():
    print(f"\n--- {name} ---")
    print("Records:", len(df))
    print("Missing values:", df.isnull().sum().sum())
    print("Duplicate IDs:", df[id_column].duplicated().sum())
    assert df[id_column].is_unique
    assert df.isnull().sum().sum() == 0
    assert df["latitude"].between(18.40, 18.65).all()
    assert df["longitude"].between(73.75, 74.00).all()

print("\nAll four datasets passed validation!")
```
